```python
import json
from pathlib import Path
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
# Пути
PROJECT_ROOT = Path.cwd().parent 
chunks_dir = PROJECT_ROOT / "data" / "chunks"
metadata_path = PROJECT_ROOT / "embeddings" / "metadata.json"
faiss_index_path = str(PROJECT_ROOT / "faiss_index")
```

```python
with open(metadata_path, "r", encoding="utf-8") as f:
    metadata = json.load(f)

texts = []
metas = []
for m in metadata:
    chunk_file = chunks_dir / m["chunk_file"]
    with open(chunk_file, "r", encoding="utf-8") as f:
        texts.append(f.read())
    metas.append(m)

print(f"Загружено {len(texts)} текстовых фрагментов.")
```

```python
embedding_model = HuggingFaceEmbeddings(model_name="multi-qa-mpnet-base-dot-v1")
```

```python
db = FAISS.from_texts(texts, embedding_model, metadatas=metas)
db.save_local(faiss_index_path)

print(f"База FAISS успешно создана и сохранена в папку: {faiss_index_path}")
```
